```python
from IPython.display import HTML, display

html = """
<!DOCTYPE html>
<html>
<head>
<style>
body {
    font-family: Arial;
    background: #f4f7fb;
    padding: 30px;
}

.container {
    max-width: 700px;
    margin: auto;
    background: white;
    padding: 25px;
    border-radius: 15px;
}

h1 {
    color: #2563eb;
}

input {
    padding: 10px;
    margin: 5px;
    border: 1px solid #ccc;
    border-radius: 8px;
}

button {
    padding: 12px 20px;
    background: #2563eb;
    color: white;
    border: none;
    border-radius: 8px;
    cursor: pointer;
}

.result {
    margin-top: 20px;
    padding: 15px;
    background: #eef6ff;
    border-radius: 10px;
}
</style>
</head>

<body>

<div class="container">

<h1> Attendance Calculator</h1>

<p>Check your subject-wise future attendance.</p>

<h3>Subject</h3>

<input id="subject" placeholder="Subject name">

<br>

<input id="attended" type="number" placeholder="Classes Attended">

<input id="delivered" type="number" placeholder="Classes Delivered">

<br>

<input id="future" type="number"
       placeholder="Future Classes You Will Attend">

<br><br>

<button onclick="calculate()">
Calculate Attendance
</button>

<div id="result" class="result"></div>

</div>

<script>

function calculate() {

    let subject =
        document.getElementById("subject").value;

    let attended =
        Number(document.getElementById("attended").value);

    let delivered =
        Number(document.getElementById("delivered").value);

    let future =
        Number(document.getElementById("future").value);


    let current =
        (attended / delivered) * 100;


    let finalAttended =
        attended + future;


    let finalDelivered =
        delivered + future;


    let finalPercentage =
        (finalAttended / finalDelivered) * 100;


    document.getElementById("result").innerHTML = `

        <h2>${subject}</h2>

        <p>
        Current Attendance:
        <b>${current.toFixed(2)}%</b>
        </p>

        <h2>
        Future Attendance:
        ${finalPercentage.toFixed(2)}%
        </h2>

        <p>
        ${finalAttended}/${finalDelivered} classes
        </p>

    `;
}

</script>

</body>
</html>
"""

display(HTML(html))
```
